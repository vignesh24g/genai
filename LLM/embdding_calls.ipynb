```python
from dotenv import load_dotenv
import os

load_dotenv()
OPENAI_API_KEY=os.getenv("OPEN_ROUTER_API_KEY")

if OPENAI_API_KEY:
    print("OPEN_ROUTER_API_KEY is set.")
else:
    print("OPEN_ROUTER_API_KEY is not set.")
```

**Naming should be what the model uses so here openai_api_key should be mentioned**

```python
from openai import OpenAI

client = OpenAI(
  base_url="https://openrouter.ai/api/v1",
  api_key=OPENAI_API_KEY,
)

embedding = client.embeddings.create(
  model="nvidia/nemotron-3-embed-1b:free",
  input="I am a data engineer",
  # input: ["text1", "text2", "text3"] # batch embeddings also supported!
  encoding_format="float"
)
embvec=embedding.data[0].embedding
print(embvec)
```
